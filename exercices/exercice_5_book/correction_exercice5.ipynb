## Énoncé

En suivant le motif `HeroBase / Hero / HeroCreate / HeroPublic` de la présentation, construisez la hiérarchie pour une entité `Book` :

- `BookBase` : champs communs `title: str` et `author: str`
- `Book` : la vraie table — ajoute `id` (clé primaire) et `isbn: str`
- `BookCreate` : schéma d'entrée API (mêmes champs que `BookBase`, plus `isbn`)
- `BookPublic` : schéma de sortie API (mêmes champs que `BookBase`, plus `id`)

````python
from sqlmodel import SQLModel, Field, Relationship
from sqlalchemy.orm import clear_mappers
````

## Correction

````python
# Ne pas toucher : permet de relancer cette cellule sans erreur.
SQLModel.metadata.clear()
clear_mappers()

# Champs communs écrits une seule fois.
class BookBase(SQLModel):
    title: str
    author: str


# La seule vraie table.
class Book(BookBase, table=True):
    id: int | None = Field(default=None, primary_key=True)
    isbn: str


# Ce que l'API reçoit : pas d'id, la base le génère.
class BookCreate(BookBase):
    isbn: str


# Ce que l'API renvoie : l'id est forcément connu.
class BookPublic(BookBase):
    id: int
````

### Explication

```
BookBase (title, author)
├── Book         + id, isbn   → la seule table
├── BookCreate   + isbn       → ce que l'API reçoit (pas d'id : la base le génère)
└── BookPublic   + id         → ce que l'API renvoie (id: int, plus optionnel)
```

Les trois classes héritent de `BookBase` : `title` et `author` ne sont écrits qu'une seule fois.
